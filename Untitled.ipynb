```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import OneHotEncoder, StandardScaler, OrdinalEncoder
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

X = df.drop(columns=["Student_ID", "Overall_Impact"])
y = df["Overall_Impact"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, random_state=42, test_size=0.2
)

numeric_columns = ["Age", "Avg_Daily_Usage_Hours", "Sleep_Hours_Per_Night", "Mental_Health_Score"]
categorical_columns = ["Gender", "Country", "Most_Used_Platform", "Affects_Academic_Performance"]
ordinal_columns = ["Academic_Level"]

difficulty_order = ["High School","Undergraduate","Graduate"]
ordinal_order = [difficulty_order]

preprocessing = ColumnTransformer(
    transformers=[
        ("Stander", StandardScaler(), numeric_columns),
        ("OHE", OneHotEncoder(handle_unknown="infrequent_if_exist", min_frequency=20), categorical_columns),
        ("Ordinal", OrdinalEncoder(categories=ordinal_order), ordinal_columns)
    ]
)

model = Pipeline(
    steps=[
        ("preprocessing", preprocessing),
        ("model", LogisticRegression())
    ]
)

param = {
    "model__C": [0.01,0.1,1,10,100],
    "model__solver": ["lbfgs"]
}

grid = GridSearchCV(model, param, cv=5, scoring="accuracy")
grid.fit(X_train,y_train)

print("Best Parameter: ", grid.best_params_)
print("\nBest Cross-val Score: \n", grid.best_score_)

best_model = grid.best_estimator_
y_pred = best_model.predict(X_test)

print("Accuracy Score: ", accuracy_score(y_test,y_pred))
print("\nClassification Report: \n", classification_report(y_test,y_pred))
print("\nConfusion Matrix: \n", confusion_matrix(y_test,y_pred))
```

```python
import joblib

countries = sorted(df["Country"].unique())
countries.remove("Other")
countries = ["Other"] + countries

options = {
    "Country": countries,
    "Gender": sorted(df["Gender"].unique()),
    "Most_Used_Platform": sorted(df["Most_Used_Platform"].unique()),
}

joblib.dump(grid.best_estimator_, "model.joblib") 
joblib.dump(options, "options.joblib")
```
